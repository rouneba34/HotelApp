```python
# ==========================================
# 🧹 NETTOYAGE + VISUALISATION AVEC PLOTLY
# ==========================================

import pandas as pd
import plotly.express as px

# 1️⃣ --- Chargement du fichier CSV ---
file_path = "airline_data.csv"  # Mets le bon chemin si besoin
df = pd.read_csv(file_path)

print("✅ Données chargées :", df.shape, "lignes et colonnes")

# 2️⃣ --- Suppression des colonnes entièrement vides ---
df = df.dropna(axis=1, how='all')

print("🧽 Colonnes après suppression des entièrement vides :", df.shape[1])

# 3️⃣ --- Identification des colonnes numériques et catégorielles ---
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# 4️⃣ --- Remplacement des valeurs manquantes ---
# Pour les colonnes numériques : médiane (ou moyenne)
for col in num_cols:
    if df[col].isna().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)

# Pour les colonnes catégorielles : mode (valeur la plus fréquente)
for col in cat_cols:
    if df[col].isna().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)

# 5️⃣ --- Suppression des doublons ---
df.drop_duplicates(inplace=True)

print("🧾 Taille finale après nettoyage :", df.shape)

# 6️⃣ --- Vérification rapide du taux de valeurs manquantes ---
missing_summary = df.isna().mean() * 100
print("\n Pourcentage de valeurs manquantes (après nettoyage) :")
print(missing_summary[missing_summary > 0].sort_values(ascending=False))

# 7️⃣ --- Création du graphique circulaire ---
# Comptage du nombre de vols par compagnie aérienne
airline_counts = df['Reporting_Airline'].value_counts().reset_index()
airline_counts.columns = ['Compagnie', 'Nombre_de_vols']

# 8️⃣ --- Graphique Plotly ---
fig = px.pie(
    airline_counts,
    names='Compagnie',
    values='Nombre_de_vols',
    title="Répartition des vols par compagnie aérienne",
    color_discrete_sequence=px.colors.qualitative.Set3,
    hole=0.4  # 👈 Pour un style donut
)

fig.update_traces(textinfo='percent+label', textposition='inside')

# 9️⃣ --- Affichage ---
fig.show()
```
